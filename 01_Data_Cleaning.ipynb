## <span style="color: #1e3a8a;">📌 What Is This Project About?</span>

<div style="background-color: #f0f4f8; padding: 15px; border-left: 4px solid #1e3a8a; border-radius: 4px;">
This project builds a <b>machine learning model</b> that predicts which UK financial firms are most likely to face <b>enforcement action from the Financial Conduct Authority (FCA)</b> in the next 24 months. Instead of predicting whether a company will go bankrupt (which is what most existing models do), this model predicts whether the regulator itself will punish the firm, through revocation, fines, or formal restrictions.
</div>

## <span style="color: #7c3aed;">🎯 Why Does This Matter?</span>

The FCA supervises approximately <b>40,000 to 50,000 firms</b> in the UK, but it has nowhere near enough inspectors to check every firm every year. In practice, the FCA can only investigate a small percentage of firms annually. This creates a critical problem:

<ul style="line-height: 1.8;">
<li><b>Which firms should the FCA look at first?</b></li>
<li><b>How can the regulator find bad actors before consumers get harmed?</b></li>
<li><b>How can limited resources produce maximum protection?</b></li>
</ul>

<div style="background-color: #fef3c7; padding: 15px; border-left: 4px solid #f59e0b; border-radius: 4px;">
Currently, the FCA relies mostly on complaints and reactive investigations. By the time they find a problem, thousands of customers may already have lost money. A <b>predictive model that ranks firms by risk</b> would allow the FCA to inspect the riskiest 10% first, turning limited inspection capacity into maximum consumer protection.
</div>

## <span style="color: #1e3a8a;">📌 How the Label (y) Was Constructed</span>

<div style="background-color: #f0f4f8; padding: 15px; border-left: 4px solid #1e3a8a; border-radius: 4px;">

### Source of the Adverse Outcome Dates

The label is built entirely from the **FCA Financial Services Register dataset** (not Companies House), using two specific fields collected during the original API sweep:

<ul style="line-height: 1.8;">
<li><b>status_date</b> — the date a firm's status changed to "Revoked", pulled from the FCA Register's main firm record (originally the API's "Status Effective Date" field)</li>
<li><b>ActionEffectiveFrom</b> — the date of each disciplinary action, pulled from the FCA Register's DisciplinaryHistory sub-endpoint (a list of dated enforcement records per firm)</li>
</ul>

Companies House data was <b>not</b> used to build the label. Companies House has no visibility into FCA regulatory actions — it only tracks incorporation, filings, directors, and dissolution. Companies House data was used separately for governance features (filing lateness, director churn) and for correcting the unrelated <code>company_status</code> leakage issue.

### The Construction Logic

For every firm-year row (a specific firm, observed on a specific date), the label was computed by working <b>backwards from the known, real-world outcome date</b> and checking whether it fell inside a 24-month forward-looking window starting from that row's observation date:

<pre style="background-color: #ffffff; padding: 10px; border-radius: 4px;">
window = (observation_date, observation_date + 24 months]

y = 1  if the firm's revocation date OR any disciplinary action date
       falls inside this window
y = 0  if no such event falls inside this window
y = censored (dropped)  if the firm voluntarily exits the Register
       inside this window, with no adverse marker
</pre>

This calculation is repeated <b>independently for every row</b>, using that row's own specific window — not a single static answer copied across a firm's whole history.

### Verification

The logic was manually verified against real firms. For example, <b>Guyon Capital Management Limited</b> (revoked 17 August 2023) produced the following labels across its 8 observation years:

<pre style="background-color: #ffffff; padding: 10px; border-radius: 4px;">
2016 window (2016-2018): revocation NOT in window → y=0
2017 window (2017-2019): revocation NOT in window → y=0
2018 window (2018-2020): revocation NOT in window → y=0
2019 window (2019-2021): revocation NOT in window → y=0
2020 window (2020-2022): revocation NOT in window → y=0
2021 window (2021-2023): revocation NOT in window → y=0
2022 window (2022-2024): revocation IS in window  → y=1
2023 window (2023-2025): revocation IS in window  → y=1
</pre>

Every row matched the expected result exactly, confirming the label construction correctly respects point-in-time discipline — each row's label depends only on that row's own forward-looking window, not on a fixed, current-state snapshot of the firm.

### Why This Differs From the Leaky Features

This same "backwards-from-outcome" logic is legitimate for the label because a dataset's ground truth must always be constructed with the benefit of hindsight (the researcher, working in 2026, necessarily knows what eventually happened). This is fundamentally different from the leakage found in <code>fca_status</code> and (originally) <code>company_status</code>: those features copied a firm's <i>current</i> value onto every historical row without any date comparison at all, whereas the label recalculates fresh for every single row against a properly bounded time window.

</div>

## <span style="color: #dc2626;">🔧 Step 3: Handling Missing Values</span>

<div style="background-color: #fef2f2; padding: 15px; border-left: 4px solid #dc2626; border-radius: 4px;">

Different columns have missing values for different reasons. A one-size-fits-all approach (like dropping all NaN rows) would destroy meaningful signal. Our strategy uses three treatments based on why data is missing:

<b>Treatment 1 — Categorical missing:</b> Replace with a "Missing" category. This preserves the information that the value was absent (which is itself predictive).

<b>Treatment 2 — Numeric with meaningful zero:</b> Create a binary flag indicating missingness, then fill with 0.

<b>Treatment 3 — Numeric with imputation:</b> Create a binary flag, then fill with the median.

<b>Why the flags matter:</b> XGBoost handles missing values natively, but logistic regression cannot. The imputation strategy preserves data for the linear baseline, while the binary flags allow both models to learn that "missingness itself is informative."

</div>

```python
# Make a copy so we don't destroy the original
df = panel.copy()

# ============ VERIFY 'frn' EXISTS ============
if 'frn' not in df.columns:
    raise ValueError("ERROR: 'frn' column missing from panel. Reload the raw data.")
print(f"Starting with {len(df.columns)} columns including 'frn'")

print("\nBEFORE handling missing values:")
before_counts = df.isnull().sum()
print(before_counts[before_counts > 0].sort_values(ascending=False))
print()

# ============ TREATMENT 1: Categorical → "Missing" category ============
print("=" * 60)
print("TREATMENT 1: Categorical columns")
print("=" * 60)
categorical_cols = ['client_money', 'company_status', 'company_type']
for col in categorical_cols:
    if col in df.columns:
        n_missing = df[col].isnull().sum()
        df[col] = df[col].fillna('Missing')
        df[col] = df[col].replace('', 'Missing')
        print(f"   {col:25s} | filled {n_missing:>6,} nulls with 'Missing'")

# ============ TREATMENT 2: Filing/churn → flag + fill with 0 ============
print("\n" + "=" * 60)
print("TREATMENT 2: Filing/churn columns (fill with 0)")
print("=" * 60)
fill_with_zero = ['accounts_days_late', 'accounts_late_3yr_avg', 'director_churn_24m']
for col in fill_with_zero:
    if col in df.columns:
        flag_col = f'{col}_missing'
        df[flag_col] = df[col].isnull().astype(int)
        n_flagged = df[flag_col].sum()
        df[col] = df[col].fillna(0)
        print(f"   {col:30s} | filled with 0 | flagged {n_flagged:>5,} rows")

# ============ TREATMENT 3: Structural → flag + median ============
print("\n" + "=" * 60)
print("TREATMENT 3: Structural columns (fill with median)")
print("=" * 60)
fill_with_median = ['director_count', 'mean_director_tenure', 'director_to_officer_ratio']
for col in fill_with_median:
    if col in df.columns:
        flag_col = f'{col}_missing'
        df[flag_col] = df[col].isnull().astype(int)
        n_flagged = df[flag_col].sum()
        median_val = df[col].median()
        df[col] = df[col].fillna(median_val)
        print(f"   {col:30s} | median={median_val:>7.2f} | flagged {n_flagged:>5,} rows")

# ============ TREATMENT 4: Within-firm imputation for firm_age_years ============
print("\n" + "=" * 60)
print("TREATMENT 4: Within-firm imputation for firm_age_years")
print("=" * 60)

# Flag missing first
df['firm_age_years_missing'] = df['firm_age_years'].isnull().astype(int)
originally_missing = df['firm_age_years_missing'].sum()

# Convert obs_date to datetime for calculations
df['_obs_dt'] = pd.to_datetime(df['obs_date'])

# For each firm, use any known age to compute missing years
def fill_age_within_firm(group):
    if group['firm_age_years'].notna().any():
        known = group[group['firm_age_years'].notna()].iloc[0]
        ref_year = known['_obs_dt'].year
        ref_age = known['firm_age_years']
        
        missing_mask = group['firm_age_years'].isnull()
        for idx in group[missing_mask].index:
            year_diff = group.loc[idx, '_obs_dt'].year - ref_year
            group.loc[idx, 'firm_age_years'] = ref_age + year_diff
    return group

# CRITICAL FIX: include_groups=False keeps 'frn' as a column after groupby
df = df.groupby('frn', group_keys=False)[df.columns].apply(fill_age_within_firm)

# Report
still_missing = df['firm_age_years'].isnull().sum()
filled_by_within_firm = originally_missing - still_missing
print(f"   Originally missing:              {originally_missing:>6,}")
print(f"   Filled by within-firm logic:     {filled_by_within_firm:>6,}")
print(f"   Still missing (no other year):   {still_missing:>6,}")

# Fallback: median for firms with no known age in any year
if still_missing > 0:
    median_age = df['firm_age_years'].median()
    df['firm_age_years'] = df['firm_age_years'].fillna(median_age)
    print(f"   Filled remainder with median:    {median_age:.1f} years")

df.drop(columns=['_obs_dt'], inplace=True)

# ============ ADDITIONAL FEATURE: has_annual_accounts ============
print("\n" + "=" * 60)
print("ADDITIONAL FEATURE: has_annual_accounts")
print("=" * 60)
df['has_annual_accounts'] = (df['accounts_days_late_missing'] == 0).astype(int)
print(f"   has_annual_accounts distribution:")
print(f"      0 (never filed AA):  {(df['has_annual_accounts']==0).sum():>6,} rows")
print(f"      1 (filed AA):        {(df['has_annual_accounts']==1).sum():>6,} rows")

# ============ VERIFY 'frn' STILL EXISTS ============
if 'frn' not in df.columns:
    print("WARNING: 'frn' column was lost during processing. Restoring from original panel.")
    df.insert(0, 'frn', panel['frn'].values)

# ============ VERIFY ============
print("\n" + "=" * 60)
print("VERIFICATION")
print("=" * 60)
remaining_nulls = df.isnull().sum()
remaining_nulls = remaining_nulls[remaining_nulls > 0]
if len(remaining_nulls) == 0:
    print("SUCCESS: No missing values remaining")
else:
    print("WARNING: Nulls still present:")
    print(remaining_nulls)

print(f"\nOriginal columns:  {len(panel.columns)}")
print(f"Cleaned columns:   {len(df.columns)}")
print(f"'frn' present?     {'frn' in df.columns}")
print(f"New columns added: {len(df.columns) - len(panel.columns)}")

new_cols = [c for c in df.columns if c not in panel.columns]
print(f"\nNew columns:")
for col in new_cols:
    print(f"   - {col}")

# ============ SAVE ============
output_path = "data/firm_year_panel_clean.csv"
df.to_csv(output_path, index=False)
print(f"\nCleaned panel saved to: {output_path}")
print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
```

## <span style="color: #dc2626;">🔧</span>

<div style="background-color: #fef2f2; padding: 15px; border-left: 4px solid #dc2626; border-radius: 4px;">

Different columns have missing values for different reasons, requiring different treatments. A blanket "fill everything with median" approach would introduce bias — for example, treating firms that never filed accounts as if they had filed median-late.

<b>Four treatments are applied based on the meaning of missingness:</b>

<b>Treatment 1 — Categorical → "Missing" category:</b> For <code>client_money</code>, <code>company_status</code>, <code>company_type</code>. The value "Missing" becomes a legitimate category the model can learn from.

<b>Treatment 2 — Filing/churn variables → Flag + fill with 0:</b> For <code>accounts_days_late</code>, <code>accounts_late_3yr_avg</code>, <code>director_churn_24m</code>. Missing here likely means "no activity", so 0 is a conservative default.

<b>Treatment 3 — Structural variables → Flag + median:</b> For <code>director_count</code>, <code>mean_director_tenure</code>, <code>director_to_officer_ratio</code>. These represent quantities that must exist but couldn't be computed.

<b>Treatment 4 — Within-firm imputation for firm_age_years:</b> If a firm has age known in ANY year, we calculate age at other years exactly using the year difference. Only firms with age missing in ALL years get median imputation.

<b>Additional feature:</b> <code>has_annual_accounts</code> — binary indicator of whether the firm has ever filed annual accounts before the observation date.

</div>

## <span style="color: #059669;">🎯 Step 4 Label Analysis: Understanding the Target Variable</span>

<div style="background-color: #f0fdf4; padding: 15px; border-left: 4px solid #059669; border-radius: 4px;">

Before analysing features, we must deeply understand the target variable <code>y</code>. This analysis reveals:

<ul style="line-height: 1.8;">
<li>The <b>rarity of adverse outcomes</b> (class imbalance severity)</li>
<li>Whether positive cases <b>cluster in specific years</b> (temporal patterns)</li>
<li>Whether positive cases cluster in certain <b>firm characteristics</b></li>

## <span style="color: #dc2626;">⚠️ Data Quality Issue: n_permissions Leakage</span>

<div style="background-color: #fef2f2; padding: 15px; border-left: 4px solid #dc2626; border-radius: 4px;">

### The Problem

During feature-label relationship analysis, an unexpected pattern emerged: 82% of adverse-outcome firms (y=1) showed <code>n_permissions = 0</code>, compared to 68% of safe firms. Investigation of individual cases revealed that firms such as <b>Ulster Bank Ltd</b> (observed in 2016, when it was a fully operational bank) showed <code>n_permissions = 0</code> in the panel — despite Ulster Bank holding dozens of active permissions at that time.

### Root Cause

The FCA Financial Services Register API provides only the <b>current</b> permission count for each firm, with no historical archive or date field attached to individual permissions. This was confirmed directly against the live API: a test query on Barclays' Permissions endpoint returned detailed permission records (Customer Type, Investment Type, Limitations) but <b>no effective date or granted date field for any individual permission</b>.

This means <code>n_permissions</code> reflects the firm's permission count <b>as of data collection (2026)</b>, not as of each historical observation date. For firms whose FCA status has since changed (approximately 20% of firm-year observations, based on a 2,000-firm sample check), this introduces temporal leakage: the feature can reflect knowledge of the firm's <i>future</i> exit from the Register.

### Verification of Alternative Sources

Three alternative data sources were investigated and ruled out:

<ol style="line-height: 1.8;">
<li><b>FCA Register Extract Service (RES)</b> — the FCA's official paid bulk data product. Documentation explicitly states that permission records "are removed after they cease to be current," confirming this source has the identical current-state-only limitation, in addition to being cost-prohibitive (indicative pricing exceeds £11,000 for a one-off extract).</li>
<li><b>Third-party scraping services</b> (e.g. Apify FCA Register Scraper) — confirmed to be built on the same official FCA Register API, and therefore subject to the same limitation.</li>
<li><b>Web archive snapshots</b> (Wayback Machine / National Archives) — the Register's firm detail pages are rendered client-side via JavaScript (Salesforce), meaning archived snapshots capture only the page shell, not the underlying permissions data. This route is additionally infeasible at the scale required (tens of thousands of firm-year observations).</li>
</ol>

<b>Conclusion: no source, free or paid, provides historical point-in-time FCA permissions data.</b> This is a genuine data availability constraint of the regulatory data ecosystem, not a project-specific limitation.

### Resolution

Rather than discard a theoretically important feature (permission scope is a well-established predictor of regulatory risk in the SupTech literature), the following transparent approach is adopted:

<ul style="line-height: 1.8;">
<li><code>n_permissions</code> and <code>has_permissions</code> are <b>retained</b> in the feature set</li>
<li>A new binary flag, <code>n_permissions_leakage_risk</code>, is added: 1 for firms whose current FCA status is not "Authorised" (indicating their permission count may reflect a post-exit state), 0 otherwise</li>
<li>A <b>sensitivity analysis</b> (Section 7.5 of the proposal) re-runs the headline model excluding <code>n_permissions</code> entirely, to assess whether the substantive conclusions depend on this feature</li>
<li>This limitation is disclosed explicitly in the dissertation's data description and limitations sections</li>
</ul>

This approach follows standard practice in empirical research: acknowledging a genuine data constraint, modelling around it transparently, and testing its impact rather than silently ignoring it or discarding a theoretically motivated feature.

</div>

## <span style="color: #f59e0b;">📐 Skewness Analysis</span>

<div style="background-color: #fef3c7; padding: 15px; border-left: 4px solid #f59e0b; border-radius: 4px;">

Skewness measures whether a feature's distribution is symmetric or lopsided, with a long tail stretching toward extreme values. This matters for two reasons:

<ul style="line-height: 1.8;">
<li><b>Logistic regression</b> assumes a roughly linear relationship between features and log-odds; heavily skewed features with extreme outliers can distort coefficient estimates</li>
<li><b>XGBoost</b> is largely robust to skewness, since it splits on thresholds rather than computing linear combinations</li>
</ul>

Features with skew > 1 are flagged as candidates for log transformation, to be applied specifically for the logistic regression baseline in the feature engineering stage.

</div>

```python
# Compute skewness for all numeric features
numeric_features = [
    'n_permissions', 'n_prior_disciplinary', 'firm_age_years',
    'accounts_days_late', 'accounts_late_3yr_avg', 'director_count',
    'director_churn_24m', 'mean_director_tenure', 'n_active_officers',
    'n_total_officers', 'director_to_officer_ratio', 'total_filings',
    'filing_count_3yr'
]

skew_data = []
for feat in numeric_features:
    skew_val = df[feat].skew()
    skew_data.append({'feature': feat, 'skew': skew_val})

skew_df = pd.DataFrame(skew_data).sort_values('skew', ascending=True)

# ============ VISUAL: Skewness bar chart ============
fig, ax = plt.subplots(figsize=(10, 7))

colors = []
for s in skew_df['skew']:
    if s > 3:
        colors.append('#dc2626')      # heavily skewed - red
    elif s > 1:
        colors.append('#f59e0b')      # moderately skewed - amber
    else:
        colors.append('#059669')      # roughly symmetric - green

bars = ax.barh(skew_df['feature'], skew_df['skew'], color=colors, alpha=0.85, edgecolor='black')

ax.axvline(x=1, color='gray', linestyle='--', alpha=0.6, label='Skew = 1 (mild threshold)')
ax.axvline(x=3, color='black', linestyle='--', alpha=0.6, label='Skew = 3 (heavy threshold)')
ax.set_xlabel('Skewness', fontsize=11)
ax.set_title('Feature Skewness (Right-Skew Severity)', fontsize=14, fontweight='bold')
ax.legend()
ax.grid(axis='x', alpha=0.3)

for bar, val in zip(bars, skew_df['skew']):
    ax.text(bar.get_width() + 0.5 if val >= 0 else bar.get_width() - 0.5,
            bar.get_y() + bar.get_height()/2,
            f'{val:.2f}', va='center', 
            ha='left' if val >= 0 else 'right', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

# ============ Text summary ============
heavy = skew_df[skew_df['skew'] > 3]['feature'].tolist()
moderate = skew_df[(skew_df['skew'] > 1) & (skew_df['skew'] <= 3)]['feature'].tolist()
symmetric = skew_df[skew_df['skew'] <= 1]['feature'].tolist()

print(f"HEAVILY SKEWED (>3) - strong candidates for log transform:")
for f in heavy:
    print(f"   {f}")

print(f"\nMODERATELY SKEWED (1-3):")
for f in moderate:
    print(f"   {f}")

print(f"\nROUGHLY SYMMETRIC (<=1) - no transform needed:")
for f in symmetric:
    print(f"   {f}")
```

### <span style="color: #1e3a8a;">Skewness Findings</span>

<div style="background-color: #f0f4f8; padding: 15px; border-left: 4px solid #1e3a8a; border-radius: 4px;">

The skewness chart confirms a clear three-tier pattern across the 13 numeric features:

<b>🔴 Heavily skewed (skew > 3) — 6 features:</b><br>
<code>accounts_days_late</code> (37.46), <code>accounts_late_3yr_avg</code> (31.66), <code>n_prior_disciplinary</code> (19.25), <code>n_active_officers</code> (6.40), <code>director_churn_24m</code> (3.74), <code>firm_age_years</code> (3.10)

These features are dominated by a large cluster of firms at or near zero, with a small number of extreme outliers stretching the distribution far to the right. This pattern is expected: filing lateness, prior disciplinary actions, and director churn are all naturally rare/small for most firms but occasionally extreme for a troubled minority.

<b>🟡 Moderately skewed (skew 1–3) — 3 features:</b><br>
<code>n_total_officers</code> (2.44), <code>director_count</code> (2.26), <code>filing_count_3yr</code> (1.86)

<b>🟢 Roughly symmetric (skew ≤ 1) — 4 features:</b><br>
<code>n_permissions</code> (0.73), <code>mean_director_tenure</code> (0.62), <code>director_to_officer_ratio</code> (−0.27), <code>total_filings</code> (−0.31)

<b>Implication for feature engineering:</b> The 6 heavily skewed features are strong candidates for log transformation (<code>log1p</code>) when preparing inputs for the L2-regularised logistic regression baseline, to reduce the influence of extreme outliers on the estimated coefficients. XGBoost will use the untransformed features directly, since tree-based splits are invariant to monotonic transformations such as log — meaning a log transform would not change how XGBoost partitions the data, but would materially affect logistic regression's linear fit.

</div>

## <span style="color: #7c3aed;">🔍 Investigation: The Client Money Paradox</span>

<div style="background-color: #f5f3ff; padding: 15px; border-left: 4px solid #7c3aed; border-radius: 4px;">

### The Concern

The categorical analysis of <code>client_money</code> produced a counter-intuitive result that warranted further investigation before being accepted at face value. The theoretical expectation, drawn from the literature and from high-profile enforcement cases cited in this project's proposal (Beaufort Securities, London Capital &amp; Finance), was that firms with full custody of client money ("Hold and control client money") would show the <b>highest</b> adverse outcome rate, since holding customer assets directly creates both the opportunity and incentive for misuse. Instead, the data showed the opposite ordering: firms classified as "Control but not hold client money" showed an adverse outcome rate of <b>1.03%</b>, nearly <b>4.5 times higher</b> than firms with full custody ("Hold and control client money" at 0.23%). Given that a similarly counter-intuitive and ultimately erroneous pattern had already been identified in the <code>n_permissions</code> feature (Section X), this result was treated as a candidate for a genuine finding only after ruling out a data leakage explanation.

### The Leakage Check

The verification procedure mirrored the one used to diagnose the <code>n_permissions</code> and <code>fca_status</code> leakage: individual firm records were pulled directly from the raw FCA JSON files and cross-referenced against their status and status-change dates. Specifically, revoked firms with a non-missing <code>client_money</code> value were manually inspected to determine whether the recorded permission type reflected the firm's <i>current</i> state (as of data collection in 2026) rather than its state at the historical observation date — the same mechanism that had corrupted <code>n_permissions</code>. Unlike <code>n_permissions</code>, which showed a firm's permission count collapsing to zero only after revocation (proving the field was not point-in-time), the <code>client_money</code> values for revoked firms were found to be either genuinely missing throughout the firm's history or consistent with a stable, unchanging permission type. No case was found in which a firm's <code>client_money</code> classification appeared to reflect a post-exit state that differed from its true historical value. This is consistent with the regulatory reality that a firm's client money permission type is typically fixed at authorisation and rarely changes over a firm's lifetime, unlike the permission count or authorisation status, both of which are mechanically altered by revocation. On this basis, <code>client_money</code> was assessed as free of the point-in-time leakage that affected <code>n_permissions</code> and <code>fca_status</code>.

### Explaining the Proportion: Why "Control but Not Hold" Shows Higher Risk

With leakage ruled out, the pattern was investigated as a genuine feature of the population. A breakdown by firm characteristics showed that firms in the "Control but not hold" category are systematically <b>older</b> (mean firm age 36.3 years, versus 27.1 years for "Hold and control") and <b>larger</b> (mean director count 6.4, versus 3.3), which initially appeared to rule out an explanation based on firm immaturity or informality. The decisive factor emerged from cross-tabulating <code>client_money</code> against <code>business_type</code>: firms with "Control but not hold" permissions are disproportionately composed of <b>overseas branches</b> (23% of the category, compared with only 0.6% for "Hold and control" firms). Overseas branches were independently identified in the business-type analysis as the highest-risk category overall (1.13% adverse rate, roughly three times the population average). The elevated rate observed for "Control but not hold" firms is therefore substantially attributable to this compositional skew toward overseas branches, rather than to the client money arrangement itself carrying greater inherent risk. A plausible complementary explanation is that firms with full custodial responsibility for client money ("Hold and control") are subject to the FCA's strictest capital adequacy and client asset safeguarding rules (the CASS regime), together with more intensive ongoing supervision, which may offset the raw riskiness of the activity through tighter regulatory oversight.

### Why This Matters for Model Design

This finding was not treated as a problem to correct but as evidence supporting the project's central methodological argument. The true driver of risk is not <code>client_money</code> or <code>business_type</code> in isolation, but their <b>interaction</b>: an overseas branch with a "Control but not hold" arrangement carries materially different risk than a UK-domiciled firm with the same permission type. Logistic regression would require this interaction to be specified manually as an explicit interaction term to capture it; XGBoost can discover such conditional relationships automatically through its tree-splitting structure. This provides direct empirical support, rather than purely theoretical justification, for the use of a nonlinear model in this project.

***The investigation showed that company_status was using today's company status for past observations, causing data leakage. Since most dissolved firms (88.6%) have a recorded cessation date, their historical status can be corrected. For the remaining 11.4% without a cessation date, the safest approach is to assume they were active, because that avoids giving the model information from the future.***

```python
CH_DIR = Path(r"E:\project\MSc-project\Data_Collection\data\raw\ch")
FCA_DIR = Path(r"E:\project\MSc-project\Data_Collection\data\raw\fca")

def parse_iso_date(s):
    if not s:
        return None
    try:
        return date.fromisoformat(str(s)[:10])
    except:
        return None

# Build frn -> cessation_date lookup (only need to do this once)
print("Building cessation date lookup...")

frn_to_ch_number = {}
for firm_file in FCA_DIR.glob("firm_*.json"):
    firm_data = json.load(open(firm_file))
    frn = firm_data.get('frn', '')
    ch_num = firm_data.get('ch_number', '').strip()
    if frn and ch_num:
        frn_to_ch_number[frn] = ch_num.zfill(8)

print(f"   Mapped {len(frn_to_ch_number)} FRNs to Companies House numbers")

cessation_lookup = {}
for ch_file in CH_DIR.glob("company_*.json"):
    ch_data = json.load(open(ch_file))
    ch_num = ch_data.get('company_number', '')
    cessation_str = ch_data.get('profile', {}).get('date_of_cessation')
    if ch_num and cessation_str:
        cessation_lookup[ch_num] = parse_iso_date(cessation_str)

print(f"   Found {len(cessation_lookup)} companies with a cessation date")
```

```python
# Map each row's frn to its cessation date (via the ch_number lookup)
df['_ch_number'] = df['frn'].astype(str).map(frn_to_ch_number)
df['_cessation_date'] = df['_ch_number'].map(cessation_lookup)

# Ensure obs_date is datetime
df['obs_date'] = pd.to_datetime(df['obs_date'])
df['_obs_date_only'] = df['obs_date'].dt.date

# Terminal statuses that need checking
terminal_statuses = ['dissolved', 'liquidation', 'administration', 'converted-closed', 'closed', 'receivership']

def fix_status(row):
    current_status = row['company_status']
    
    # If not a terminal status, keep as-is (no leakage risk)
    if current_status not in terminal_statuses:
        return current_status
    
    cessation = row['_cessation_date']
    obs = row['_obs_date_only']
    
    if pd.isna(cessation):
        # No cessation date available - safer default assumption
        return 'active'
    
    if obs < cessation:
        # Firm was STILL ALIVE at the observation date
        return 'active'
    else:
        # Firm had genuinely already ceased by the observation date
        return current_status

print("Applying point-in-time fix to company_status...")
df['company_status_fixed'] = df.apply(fix_status, axis=1)

# Cleanup temp columns
df = df.drop(columns=['_ch_number', '_cessation_date', '_obs_date_only'])

# ============ VERIFY THE FIX ============
print("\nBEFORE fix (leaky, current-state):")
print(df['company_status'].value_counts())

print("\nAFTER fix (point-in-time corrected):")
print(df['company_status_fixed'].value_counts())

# Spot check the exact firms we found leaking earlier
print("\n\nSpot-check on known leaky firms:")
check_frns = ['161302', '184115', '197332']
for frn in check_frns:
    firm_rows = df[df['frn'].astype(str) == frn][['obs_date', 'company_status', 'company_status_fixed']]
    print(f"\nFRN {frn}:")
    print(firm_rows.to_string(index=False))
```

### <span style="color: #1e3a8a;">How the Correction Was Applied</span>

<div style="background-color: #f0f4f8; padding: 15px; border-left: 4px solid #1e3a8a; border-radius: 4px;">

For each firm-year row, the correction compared the observation date against the firm's actual <code>date_of_cessation</code> recorded in Companies House:

<ul style="line-height: 1.8;">
<li>If the current status was <b>not</b> a terminal state (e.g. "active", "open") — no change was needed, since there is no future-state information to leak</li>
<li>If the current status <b>was</b> a terminal state (dissolved, liquidation, administration, etc.) <b>and</b> a cessation date was available — the observation date was compared against it: if the observation occurred <b>before</b> the cessation date, the firm was still alive at that point and the status was corrected to "active"; if the observation occurred <b>on or after</b> the cessation date, the terminal status was genuinely correct and left unchanged</li>
<li>If the current status was terminal <b>but no cessation date was recorded</b> (~11% of cases) — the status was defaulted to "active", the safer assumption given most such firms were still operating at the time</li>
</ul>

This reconstructs what a real FCA supervisor would have observed at each historical point in time, using only information that predates the observation date — eliminating the leakage caused by applying the firm's 2026 current-state label uniformly across its entire panel history.

## <span style="color: #059669;">✅ Data Quality Issue: fca_status Leakage — Fixed and Verified</span>

<div style="background-color: #f0fdf4; padding: 15px; border-left: 4px solid #059669; border-radius: 4px;">

### The Problem

Firms currently labelled <code>fca_status = "Revoked"</code> showed a 36.08% adverse outcome rate — about 100 times higher than the dataset average (0.37%). This happened because <code>fca_status</code> was pulled from each firm's <b>current</b> status (as of data collection in 2026) and applied identically to every historical row, regardless of when the observation actually took place. A firm's 2016 row could show "Revoked" even if the real revocation only happened in 2023, seven years later.

### The Fix

Each row's <code>fca_status</code> was recalculated using the firm's actual <code>status_date</code>:

<ul style="line-height: 1.8;">
<li>If the observation date came <b>before</b> the firm's real status change date, the row was corrected to <b>"Authorised"</b>, since that reflects what was true at that point in time</li>
<li>If the observation date came <b>on or after</b> the real status change date, the existing label was left as-is (already correct)</li>
</ul>

### What Changed After the Fix

The "Revoked" category disappeared entirely from <code>fca_status</code>, and this is the <b>correct, expected result</b> — not a mistake or data loss. Firms only stop appearing in the panel once they are actually revoked, so no row in the dataset ever has an observation date on or after its own firm's true revocation date. Every row that previously said "Revoked" has now been correctly reassigned to "Authorised", since those observations genuinely occurred while the firm was still active.

<b>The label (<code>y</code>) was never touched by this fix.</b> The label is built independently from the raw revocation and disciplinary dates, compared against each row's own 24-month window — not from the <code>fca_status</code> text field. Verification confirms all 307 positive cases remain fully intact: for example, Guyon Capital Management (revoked August 2023) still correctly shows <code>y=1</code> for its 2022 and 2023 rows, even though <code>fca_status</code> now correctly shows "Authorised" for those same rows, since the firm was indeed still authorised at those observation dates — only the future 24-month window captured the upcoming revocation.

```python
# Recreate the BEFORE data (from our investigation earlier)
before_data = pd.DataFrame({
    'status': ['Revoked', 'Recognised', 'No longer authorised', 'Authorised', 
               'See full details', 'Authorised - applied to cancel', 
               'Authorised Schedule 5', 'EEA Authorised', 'Registered'],
    'rate_pct': [36.08, 2.78, 0.26, 0.21, 0.21, 0.17, 0.00, 0.00, 0.00]
})

# Current AFTER data (from the fixed df)
after_data = df.groupby('fca_status').agg(total=('y','size'), positive=('y','sum')).reset_index()
after_data['rate_pct'] = (after_data['positive']/after_data['total']*100).round(2)
after_data = after_data.rename(columns={'fca_status': 'status'})

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# BEFORE plot
ax = axes[0]
before_sorted = before_data.sort_values('rate_pct', ascending=True)
colors_before = ['#dc2626' if r > 5 else '#059669' for r in before_sorted['rate_pct']]
ax.barh(before_sorted['status'], before_sorted['rate_pct'], color=colors_before, alpha=0.85, edgecolor='black')
ax.set_xlabel('Adverse outcome rate (%)', fontsize=11)
ax.set_title('BEFORE Fix\n(Revoked = 36.08% - the leak)', fontsize=13, fontweight='bold', color='#dc2626')
ax.grid(axis='x', alpha=0.3)
for i, (status, rate) in enumerate(zip(before_sorted['status'], before_sorted['rate_pct'])):
    ax.text(rate + 0.5, i, f'{rate:.2f}%', va='center', fontsize=9, fontweight='bold')

# AFTER plot
ax = axes[1]
after_sorted = after_data.sort_values('rate_pct', ascending=True)
colors_after = ['#059669'] * len(after_sorted)
ax.barh(after_sorted['status'], after_sorted['rate_pct'], color=colors_after, alpha=0.85, edgecolor='black')
ax.set_xlabel('Adverse outcome rate (%)', fontsize=11)
ax.set_title('AFTER Fix\n("Revoked" spike eliminated)', fontsize=13, fontweight='bold', color='#059669')
ax.set_xlim(axes[0].get_xlim())  # same scale for fair comparison
ax.grid(axis='x', alpha=0.3)
for i, (status, rate) in enumerate(zip(after_sorted['status'], after_sorted['rate_pct'])):
    ax.text(rate + 0.5, i, f'{rate:.2f}%', va='center', fontsize=9, fontweight='bold')

plt.suptitle('fca_status Leakage: Before vs After Point-in-Time Correction', 
             fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## <span style="color: #1e3a8a;">📝 Summary of Key Insights</span>

<div style="background-color: #f0f4f8; padding: 15px; border-left: 4px solid #1e3a8a; border-radius: 4px;">

Bringing together everything explored in this notebook, seven main patterns stand out:

<ol style="line-height: 2;">

<li><b>Firms that have filed late before tend to keep filing late.</b> The strong link between <code>accounts_days_late</code> and <code>accounts_late_3yr_avg</code> shows that filing lateness is not usually a one-off mistake — it's a habit that persists year after year.</li>

<li><b>Firms that file more often tend to be better behaved.</b> Higher recent filing activity (<code>filing_count_3yr</code>) is associated with a lower chance of adverse outcomes, suggesting that firms which stay active and engaged with their statutory paperwork are generally easier to trust.</li>

<li><b>Firms losing directors quickly are more likely to run into trouble.</b> Higher director churn (<code>director_churn_24m</code>) is linked to a higher chance of adverse outcomes — when people leave a firm's leadership rapidly, it often signals internal problems building up.</li>

<li><b>A firm's past disciplinary record is rare but says a lot.</b> Very few firms have any prior disciplinary history, but when they do, it is one of the strongest warning signs in the entire dataset.</li>

<li><b>Most firms look perfectly normal — a small few stand out.</b> The vast majority of firms show low, unremarkable values across every feature. Only a small number show extreme values (very late filings, very high director churn, multiple past disciplinary actions), and these extreme firms are the ones worth paying closer attention to.</li>

<li><b>No single number tells the whole story.</b> Safe and adverse firms overlap heavily on almost every individual feature. This means no single warning sign is enough on its own to predict trouble — the model needs to combine several pieces of evidence together to make a confident prediction, which is exactly why a model like XGBoost (which can weigh multiple signals at once) is expected to outperform relying on any one feature alone.</li>

<li><b>Extreme behaviour is more common among troubled firms, but not universal.</b> While firms heading toward regulatory trouble are more likely to show extreme values, many still look ordinary right up until something happens — a reminder that regulatory risk is genuinely hard to predict and why this project's evaluation approach (precision at top decile, recall, calibration) matters as much as the model itself.</li>

</ol>

</div>
